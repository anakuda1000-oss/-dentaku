```python
import ipywidgets as widgets
from IPython.display import display, clear_output

# 表示用のテキストボックス
output_text = widgets.Text(
    value='',
    placeholder='',
    description='',
    disabled=True,
    layout=widgets.Layout(width='230px', margin='10px')
)

expression = ""

def on_button_clicked(b):
    global expression
    val = b.description

    if val == '=':
        try:
            expression = str(eval(expression))
        except ZeroDivisionError:
            expression = "Error (0割不可)"
        except Exception:
            expression = "Error"
    elif val == 'C':
        expression = ""
    else:
        expression += val

    output_text.value = expression

# ボタンの配置設定
buttons_layout = [
    ['7', '8', '9', '/'],
    ['4', '5', '6', '*'],
    ['1', '2', '3', '-'],
    ['0', '.', '+', '='],
    ['C']
]

# UIの組み立て
rows = [output_text]
for row_data in buttons_layout:
    row_buttons = []
    for char in row_data:
        # Cボタンのみ横幅を広げる
        btn_width = '230px' if char == 'C' else '50px'
        btn = widgets.Button(description=char, layout=widgets.Layout(width=btn_width, margin='2px'))
        btn.on_click(on_button_clicked)
        row_buttons.append(btn)
    rows.append(widgets.HBox(row_buttons))

# 縦に並べて表示
calculator_ui = widgets.VBox(rows)
display(calculator_ui)
```
