```python
import tkinter as tk

class Calculator:
    def __init__(self, root):
        self.root = root
        self.root.title("電卓")
        self.root.geometry("300x400")

        self.expression = ""

        # ディスプレイ表示エリア
        self.display = tk.Entry(root, font=("Arial", 20), borderwidth=5, relief=tk.RIDGE, justify="right")
        self.display.grid(row=0, column=0, columnspan=4, padx=10, pady=10, ipady=10)

        # ボタンの配置設定
        buttons = [
            ('7', 1, 0), ('8', 1, 1), ('9', 1, 2), ('/', 1, 3),
            ('4', 2, 0), ('5', 2, 1), ('6', 2, 2), ('*', 2, 3),
            ('1', 3, 0), ('2', 3, 1), ('3', 3, 2), ('-', 3, 3),
            ('0', 4, 0), ('.', 4, 1), ('+', 4, 2), ('=', 4, 3),
            ('C', 5, 0)
        ]

        # ボタンの生成
        for text, row, col in buttons:
            if text == '=':
                btn = tk.Button(root, text=text, font=("Arial", 14), command=self.calculate)
            elif text == 'C':
                btn = tk.Button(root, text=text, font=("Arial", 14), command=self.clear)
            else:
                btn = tk.Button(root, text=text, font=("Arial", 14), command=lambda t=text: self.on_click(t))

            # Cボタンのみ横幅を広く設定
            if text == 'C':
                btn.grid(row=row, column=col, columnspan=4, sticky="nsew", padx=5, pady=5)
            else:
                btn.grid(row=row, column=col, sticky="nsew", padx=5, pady=5)

        # グリッドの伸縮設定
        for i in range(6):
            root.grid_rowconfigure(i, weight=1)
        for i in range(4):
            root.grid_columnconfigure(i, weight=1)

    def on_click(self, char):
        self.expression += str(char)
        self.update_display()

    def clear(self):
        self.expression = ""
        self.update_display()

    def calculate(self):
        try:
            # 入力された数式を評価
            result = str(eval(self.expression))
            self.expression = result
        except ZeroDivisionError:
            self.expression = "Error (0割不可)"
        except Exception:
            self.expression = "Error"
        self.update_display()

    def update_display(self):
        self.display.delete(0, tk.END)
        self.display.insert(0, self.expression)

if __name__ == "__main__":
    root = tk.Tk()
    app = Calculator(root)
    root.mainloop()
```
